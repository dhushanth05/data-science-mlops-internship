```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x =[1,2,3,4,7]
y =[2,4,6,5,10]
plt.figure(figsize= (8,5))
plt.plot(x,y)
plt.plot(y,x)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.title('Simple line plot')
```

```python
x = np.linspace(0,10,100)
y = np.sin(x)

plt.figure(figsize= (10,5))
plt.plot(x,y)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.title('sine wave')
```

```python
x = np.linspace(0,10,100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)

plt.figure(figsize= (10,5))
plt.plot(x,y1, label= 'sin(x)')
plt.plot(x,y2, label= 'cos(x)')
plt.plot(x,y3, label= 'sin(x)*cos(x)')
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.grid(True)
plt.title('multiple function')
```

```python
x = np.linspace(0,10,20)

plt.figure(figsize= (10,5))
plt.plot(x,x+1, 'b-', label= 'a')
plt.plot(x,x+2, 'm--', label= 'b')
plt.plot(x,x+3, 'r-.', label= 'c')
plt.plot(x,x+4, 'k:', label= 'd')
plt.plot(x,x+5, 'm*--', label= 'e')
plt.plot(x,x+6, 'yo-.', label= 'f')
plt.plot(x,x+7, 'gs--', label= 'h')
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.grid(True)
plt.title('multiple function')
```

```python
np.random.seed(42)
x = np.random.randn(100)
y = 2 * x + np.random.rand(100) * 0.5

plt.figure(figsize=(8,6))
plt.scatter(x,y,alpha=0.6)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True,alpha=0.3)
plt.title('scatter plot')
```

```python
np.random.seed(42)
x = np.random.randn(100)
y = np.random.randn(100)
colour = np.random.randn(100)
size = 1000 * np.random.rand(100) 

plt.figure(figsize=(8,6))
plt.scatter(x,y, c=colour, s=size,alpha=0.6)
plt.colorbar()
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True,alpha=0.3)
plt.title('scatter plot')
```

```python
x =[1,2,3,4,5]
y =[2,4,6,5,10]
plt.figure(figsize= (8,5))
plt.bar(x,y,color='steelblue',edgecolor='black',linewidth=1.2)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True,alpha=0.3,axis='y')
plt.title('bar plot')
```

```python

plt.figure(figsize= (8,5))
plt.barh(x,y,color='steelblue',edgecolor='black',linewidth=1.2)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title('bar plot')
plt.grid(True,alpha=0.3)
```

```python
ca = ['q1','q2','q3','q4']
a = [20,30,40,25]
b = [32,25,20,30]
c = [26,19,32,29]

x = np.arange(len(ca))
width = 0.25
plt.figure(figsize=(10,6))
plt.bar(x-width,a,width,label='a')
plt.bar(x,b,width,label='b')
plt.bar(x+width,c,width,label='c')
plt.xticks(x,ca)
plt.legend()
```

```python
np.random.seed(42)
x = np.random.normal(100,15,1000)

plt.figure(figsize=(8,6))
plt.hist(x,bins=30,color='skyblue',edgecolor='black',alpha=0.6)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True,alpha=0.3)
plt.title('histogram')
```

```python
np.random.seed(42)
x = np.random.normal(100,15,1000)
y = np.random.normal(120,20,1000)
plt.figure(figsize=(8,6))
plt.hist(x,bins=30,color='skyblue',alpha=0.6)
plt.hist(y,bins=30,color='red',alpha=0.6)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True,alpha=0.3)
plt.legend()
plt.title('histogram')
```

```python
data = pd.DataFrame({
    'Name': ['Amit', 'Rahul', 'Priya', 'Sneha', 'Arjun',
             'Neha', 'Vikram', 'Anjali', 'Kiran', 'Riya'],
    'Age': [22, 25, 21, 24, 23, 26, 28, 22, 27, 24],
    'City': ['Delhi', 'Mumbai', 'Chennai', 'Bangalore', 'Hyderabad',
             'Pune', 'Delhi', 'Mumbai', 'Chennai', 'Pune'],
    'Salary': [30000, 45000, 28000, 50000, 35000,
               60000, 55000, 32000, 48000, 40000],
    'Experience': [1, 3, 0, 4, 2, 6, 5, 1, 4, 3]
})

data
```

```python
plt.figure(figsize=(10,6))
plt.plot(data['Name'],data['Salary'],label="a",linewidth=2)
plt.xticks(rotation=45)
```

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 10))
axes[0].bar(data['Name'], data['Salary'],color='skyblue', edgecolor='black')
axes[0].set_title('Bar Graph')
axes[1].scatter(data['Age'], data['Experience'])
axes[1].set_title('Age vs Experience')
plt.tight_layout()
plt.show()
fig.savefig('my_plot.png',dpi=300)
```
