```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName('practice1').getOrCreate()
```

```python
df = spark.read.csv('malaysia_aml_fraud_anomaly_transactions.csv',header=True,inferSchema=True)
```

```python
from pyspark.sql.functions import *
```

```python
df.select(sum(col("customer_age").isNull().cast("int"))).show()
```

```python
df.select([sum(col(c).isNull().cast("int")).alias(c) for c in df.columns]).show(vertical=True)
```

```python
df = df.fillna({'occupation':'Unknown','kyc_status':'Unknown','merchant_category': 'Unknown','device_type': 'Unknown','ip_country': 'Unknown','customer_message':'Unknown'})
```

```python
df.select([sum(col(c).isNull().cast("int")).alias(c) for c in df.columns]).show(vertical=True)
```

```python
dup_count = df.count() - df.dropDuplicates().count()
print(dup_count)
```

```python
df.printSchema()
```

```python
df = df.withColumn('balance_change_myr',col('balance_before_myr') - col('balance_after_myr'))
```

```python
df.select(['transaction_amount_myr','balance_change_myr','balance_before_myr','balance_after_myr']).show(20)
```

```python
df = df.withColumn('transaction_size',when(col('transaction_amount_myr')<1000,'Low').when(col('transaction_amount_myr')<5000,'Medium').otherwise('High'))
```

```python
df.select('transaction_amount_myr','transaction_size').show()
```

```python
df.groupBy("risk_category").count().show()
```

```python
df.groupBy("risk_category").agg(avg("transaction_amount_myr").alias("average_transaction")).show()
```

```python
df.groupBy("risk_category").agg(count("*").alias("transaction_count"),
    avg("transaction_amount_myr").alias("average_amount"),
    max("transaction_amount_myr").alias("maximum_amount"),
    min("transaction_amount_myr").alias("minimum_amount")
).show()
```

```python
df.filter((col("risk_category") == "High") & (col("transaction_amount_myr") > 5000)).show()
```

```python
df.filter(col("risk_category") == "High").count()
```

```python
high_risk_df = df.filter(col('risk_category') == 'High')
```

```python
high_risk_df.show()
```
