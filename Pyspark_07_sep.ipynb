```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName('practice').getOrCreate()
```

```python
spark
```

```python
df_pyspark = spark.read.csv('test2.csv',header=True,inferSchema=True)
```

```python
df_pyspark.show()
```

```python
df_pyspark.na.drop().show()
```

```python
df_pyspark.na.drop(how="any",thresh=2).show()
```

```python
df_pyspark.na.drop(subset=['Name']).show()
```

```python
df_pyspark.na.fill('missing').show()
```

```python
df_pyspark.fillna(0).show()
```

```python
df_pyspark.fillna({'Name': 'missing','age': 0,'Experience': 0,'Salary': 0}).show()
```

```python
from pyspark.ml.feature import Imputer
imputer = Imputer(inputCols=['age','Experience','Salary'],
                  outputCols=["{}_imputed".format(c)
                              for c in ['age','Experience','Salary']]).setStrategy("mean")
```

```python
imputer.fit(df_pyspark).transform(df_pyspark).show()
```
